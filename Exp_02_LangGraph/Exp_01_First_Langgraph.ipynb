```python
from os import name

from langchain_core import messages
from langgraph import graph
%load_ext dotenv
%dotenv
%load_ext mypy-ipython
```

```python
from langgraph.graph import (
    START,
    END,
    StateGraph
)
from langchain_openai.chat_models import  ChatOpenAI
from langchain_core.runnables import Runnable
from langchain_core.messages import HumanMessage , BaseMessage
from typing_extensions import  TypedDict
from collections.abc import  Sequence
```

```python
class State(TypedDict):
    message : Sequence[BaseMessage]
```

```python
state = State(message= [
    HumanMessage('''
        Best Programing Language for AI
    ''')
])
```

```python
state['message'][0].pretty_print()
```

```python
chat = ChatOpenAI(
    model='gpt-4o',
    seed=365,
    temperature=0.9,
    max_tokens=500,
)
```

```python
def chat_bot(state: State) -> State:
    print(f"\n------------chat_bot  node working")
    response = chat.invoke(state=state['message'])
    response.pretty_print()
    return  State(messages = [response])
```

```python
chat_bot(state=state)
```

```python
graph = StateGraph(State)
```

```python
graph.add_node('Chat Bot', chat_bot)
graph.add_edge(START, 'Chat Bot')
graph.add_edge('Chat Bot', END)
```

```python
graph_compiled = graph.compile()
```

```python
graph_compiled
```

```python
graph_compiled.invoke(state)
```
