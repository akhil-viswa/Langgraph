```python
import pinecone
from pathspec.patterns.gitignore import spec
from pinecone import Pinecone , ServerlessSpec
import os
from dotenv import load_dotenv , find_dotenv
from prompt_toolkit.layout import dimension
```

```python
load_dotenv(find_dotenv() , override=True)
```

```python
pc = Pinecone(
    api_key=os.getenv("PINECONE_API_KEY")
)
```

```python
pc.list_indexes()
```

```python
index_name = 'my-index'
dimension = 3
metric = 'cosine'
```

```python
if index_name in [index.name for index in pc.list_indexes()]:
    pc.delete_index(index_name)
```

```python
pc.list_indexes()
```

```python
pc.create_index(
    name=index_name,
    dimension=dimension,
    metric=metric,
    spec = ServerlessSpec(
        cloud='aws',
        region='us-east-1'
    )
)
```

```python
pc.list_indexes()
```
