```python
import numpy as np
import  matplotlib.pyplot as plt

import  librosa
import  librosa.display
import soundfile as sf
import  speech_recognition as sr

from jiwer import wer , cer
from IPython.display import Audio
```

```python
audio_single , sample_rate = librosa.load('speech_01.wav',sr=None)
```

```python
plt.figure(figsize=(10,10))
librosa.display.waveshow(audio_single,sr=sample_rate)
plt.title('Audio Waveform')
plt.show()

Audio('speech_01.wav')
```

```python
s = librosa.stft(audio_single)
```

```python
s_db = librosa.amplitude_to_db(abs(s) , ref=np.max)
```

```python
np.max(s_db)
```

```python
plt.figure(figsize=(12,5))
librosa.display.specshow(
    data=s_db,
    sr=sample_rate,
    x_axis='time',
    y_axis='log',
)
plt.colorbar(format='%+2.0f dB')
plt.show()
```

```python
signal_fil = librosa.effects.preemphasis(
    audio_single,
    coef=0.97,
)

sf.write('filtered_speach_01.wav', signal_fil, sample_rate)
```

```python
Audio('filtered_speach_01.wav')
```

```python



s = librosa.stft(signal_fil)
s_db = librosa.amplitude_to_db(abs(s) , ref=np.max)
np.max(s_db)
```
